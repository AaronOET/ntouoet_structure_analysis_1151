# 斜率與撓度 1 — 簡支梁受集中載重（互動版）

簡支梁：$A$ 為鉸支承、$B$ 為滾支承，$L = 10$ m，$EI$ 為常數。集中載重 $P$（向下）位於距 $A$ 為 $a$ 處，$b = L - a$。

反力：$A_y = \dfrac{Pb}{L}$，$B_y = \dfrac{Pa}{L}$。以 Macaulay 函數寫出彎矩並積分兩次：

$$EI\,\frac{d^2v}{dx^2} = A_y\,x - P\langle x - a\rangle$$

$$EI\,\theta = \frac{A_y}{2}x^2 - \frac{P}{2}\langle x - a\rangle^2 + C_1,\qquad
EI\,v = \frac{A_y}{6}x^3 - \frac{P}{6}\langle x - a\rangle^3 + C_1 x + C_2$$

邊界條件 $v(0) = 0$、$v(L) = 0$ 求得 $C_2 = 0$、$C_1 = -\dfrac{Pb\,(L^2 - b^2)}{6L}$。常用結果：

- 載重點撓度 $v_a = -\dfrac{P a^2 b^2}{3EIL}$；載重在跨中時 $v_{\max} = -\dfrac{PL^3}{48EI}$
- 端點斜率 $\theta_A = -\dfrac{Pb(L^2 - b^2)}{6EIL}$，$\theta_B = +\dfrac{Pa(L^2 - a^2)}{6EIL}$

**操作方式**
- 拖曳 **載重位置 a** 滑桿或直接輸入 $a$，移動集中載重 $P$，觀察彈性曲線、$\theta(x)$ 與 $v(x)$ 如何變化。
- 拖曳 **觀察點 x** 讀取該點的斜率 $\theta$ 與撓度 $v$（橘色方塊）。
- 可修改 $P$ 與 $EI$；縱軸範圍固定為載重走遍全梁時的最大值，方便比較不同載重位置。
- 勾選 **顯示計算過程** 可看到以 Macaulay 函數積分的每一步，並與課本公式比對。

符號規定：$v$ 向上為正（向下撓度為負），$\theta = dv/dx$ 逆時針為正，正彎矩使梁下緣受拉；$EI\,\dfrac{d^2v}{dx^2} = M(x)$。
