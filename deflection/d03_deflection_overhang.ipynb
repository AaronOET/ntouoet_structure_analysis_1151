# 斜率與撓度 3 — 外伸梁受集中載重（互動版）

外伸梁：$A$ 為鉸支承、$B$ 為滾支承，$L_{AB} = 8$ m，$B$ 右側外伸 4 m（全長 12 m），$EI$ 為常數。集中載重 $P$（向下）位於距 $A$ 為 $a$ 處（$0 \le a \le 12$）。

反力：$B_y = \dfrac{Pa}{L_{AB}}$，$A_y = P - B_y$（載重在外伸段時 $A_y < 0$，即向下）。彎矩與積分：

$$EI\,\frac{d^2v}{dx^2} = A_y\,x - P\langle x - a\rangle + B_y\langle x - L_{AB}\rangle$$

$$EI\,v = \frac{A_y}{6}x^3 - \frac{P}{6}\langle x - a\rangle^3 + \frac{B_y}{6}\langle x - L_{AB}\rangle^3 + C_1 x + C_2$$

邊界條件 $v(0) = 0$、$v(L_{AB}) = 0$ 求 $C_1$、$C_2$。常用結果：

- 載重在 $AB$ 跨內：跨內與簡支梁相同，$v_a = -\dfrac{Pa^2b^2}{3EIL_{AB}}$（$b = L_{AB} - a$），外伸段以直線**向上**翹起。
- 載重在外伸段、距 $B$ 為 $c$：$v_a = -\dfrac{Pc^2(L_{AB} + c)}{3EI}$，$AB$ 跨內**向上**拱起。

**操作方式**
- 拖曳 **載重位置 a** 滑桿或直接輸入 $a$，移動集中載重 $P$，觀察彈性曲線、$\theta(x)$ 與 $v(x)$ 如何變化。
- 拖曳 **觀察點 x** 讀取該點的斜率 $\theta$ 與撓度 $v$（橘色方塊）。
- 可修改 $P$ 與 $EI$；縱軸範圍固定為載重走遍全梁時的最大值，方便比較不同載重位置。
- 勾選 **顯示計算過程** 可看到以 Macaulay 函數積分的每一步，並與課本公式比對。

符號規定：$v$ 向上為正（向下撓度為負），$\theta = dv/dx$ 逆時針為正，正彎矩使梁下緣受拉；$EI\,\dfrac{d^2v}{dx^2} = M(x)$。
