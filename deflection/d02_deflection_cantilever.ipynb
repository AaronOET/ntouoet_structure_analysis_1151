# 斜率與撓度 2 — 懸臂梁受集中載重（互動版）

懸臂梁：$A$ 為固定端，右端自由，$L = 6$ m，$EI$ 為常數。集中載重 $P$（向下）位於距 $A$ 為 $a$ 處。

反力：$A_y = P$，$M_A = Pa$（逆時針）。彎矩與積分：

$$EI\,\frac{d^2v}{dx^2} = -Pa + P\,x - P\langle x - a\rangle$$

$$EI\,\theta = -Pa\,x + \frac{P}{2}x^2 - \frac{P}{2}\langle x - a\rangle^2 + C_1,\qquad
EI\,v = -\frac{Pa}{2}x^2 + \frac{P}{6}x^3 - \frac{P}{6}\langle x - a\rangle^3 + C_1 x + C_2$$

固定端的邊界條件 $\theta(0) = 0$、$v(0) = 0$ 得 $C_1 = C_2 = 0$。常用結果：

- 載重點：$\theta_a = -\dfrac{Pa^2}{2EI}$，$v_a = -\dfrac{Pa^3}{3EI}$
- 自由端：$\theta_{\text{自由端}} = \theta_a$，$v_{\text{自由端}} = -\dfrac{Pa^2(3L - a)}{6EI}$；載重在自由端時 $v = -\dfrac{PL^3}{3EI}$

**操作方式**
- 拖曳 **載重位置 a** 滑桿或直接輸入 $a$，移動集中載重 $P$，觀察彈性曲線、$\theta(x)$ 與 $v(x)$ 如何變化。
- 拖曳 **觀察點 x** 讀取該點的斜率 $\theta$ 與撓度 $v$（橘色方塊）。
- 可修改 $P$ 與 $EI$；縱軸範圍固定為載重走遍全梁時的最大值，方便比較不同載重位置。
- 勾選 **顯示計算過程** 可看到以 Macaulay 函數積分的每一步，並與課本公式比對。

符號規定：$v$ 向上為正（向下撓度為負），$\theta = dv/dx$ 逆時針為正，正彎矩使梁下緣受拉；$EI\,\dfrac{d^2v}{dx^2} = M(x)$。
