# 例題 6.4 — C 點剪力的影響線（外伸梁，互動版）

外伸梁：$A$ 為鉸支承、$B$ 為滾支承，$AB = 8$ m，$B$ 右側外伸 4 m；$C$ 點距 $A$ 為 4 m。單位載重（1）位於距 $A$ 為 $x$ 處。

先由 $\Sigma M_A = 0$ 求 $B_y = \tfrac{1}{8}x$，再於 $C$ 點切開、取右段（含 $B$ 與外伸段）：

$0 \le x < 4$（載重在 $AC$ 上）：$\;V_C = -\tfrac{1}{8}x$

$4 < x \le 12$（載重在 $C$ 右側）：$\;V_C = 1 - \tfrac{1}{8}x$

**操作方式**
- 拖曳滑桿或直接輸入 $x$ 來移動單位載重。
- 剪力在 $C$ 點**不連續**：當 $x = 4$ 時，用 **C 左側 4⁻ / C 右側 4⁺** 切換載重位於 $C$ 的緊鄰左側或右側。
- 按 **記錄此點** 逐點描出影響線；勾選 **顯示完整影響線** 對照理論結果。
