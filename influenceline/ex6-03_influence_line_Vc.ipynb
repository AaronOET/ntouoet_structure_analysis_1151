# 例題 6.3 — C 點剪力的影響線（互動版）

簡支梁 $AB$，跨距 10 m，$C$ 點距 $A$ 為 2.5 m。單位載重（1）位於距 $A$ 為 $x$ 處。

先由 $\Sigma M_A = 0$ 求 $B_y = \tfrac{1}{10}x$，再於 $C$ 點切開、取右段 $CB$：

$0 \le x < 2.5$（載重在 $AC$ 上）：$\;V_C = -\tfrac{1}{10}x$

$2.5 < x \le 10$（載重在 $CB$ 上）：$\;V_C = 1 - \tfrac{1}{10}x$

**操作方式**
- 拖曳滑桿或直接輸入 $x$ 來移動單位載重。
- 剪力在 $C$ 點**不連續**：當 $x = 2.5$ 時，用 **C 左側 2.5⁻ / C 右側 2.5⁺** 切換載重位於 $C$ 的緊鄰左側或右側。
- 按 **記錄此點** 逐點描出影響線；勾選 **顯示完整影響線** 對照理論結果。
